```python
  #1. Arithmetic Operators
  #Used to perform standard mathematical calculations.

  #┌──────────┬────────────────┬──────────────────────────────────────────────────────┬─────────┬────────┐
  #│ Operator │ Name           │ Simple Definition                                    │ Example │ Result │
  #├──────────┼────────────────┼──────────────────────────────────────────────────────┼─────────┼────────┤
  #│ +        │ Addition       │ Adds two numbers                                     │ 5 + 3   │ 8      │
  #│ -        │ Subtraction    │ Subtracts the second number from the first           │ 5 - 3   │ 2      │
  #│ *        │ Multiplication │ Multiplies two numbers                               │ 5 * 3   │ 15     │
  #│ /        │ Division       │ Divides and always gives a decimal (float) value     │ 5 / 2   │ 2.5    │
  #│ %        │ Modulus        │ Divides and gives the remainder                      │ 5 % 2   │ 1      │
  #│ //       │ Floor Division │ Divides and rounds down to the nearest whole integer │ 5 // 2  │ 2      │
  #│ **       │ Exponentiation │ Calculates power (raised to the power of)            │ 5 ** 2  │ 25     │
  #└──────────┴────────────────┴──────────────────────────────────────────────────────┴─────────┴────────┘
```

```python
#  2. Relational (Comparison) Operators
  #Used to compare two values. The result is always a boolean value: either True or False.

  #┌──────────┬──────────────────────────┬────────────────────────────────────┬─────────┬────────┐
  #│ Operator │ Name                     │ Simple Definition                  │ Example │ Result │
  #├──────────┼──────────────────────────┼────────────────────────────────────┼─────────┼────────┤
  #│ ==       │ Equal to                 │ Checks if two values are equal     │ 5 == 5  │ True   │
  #│ !=       │ Not equal to             │ Checks if two values are different │ 5 != 3  │ True   │
  #│ >        │ Greater than             │ Checks if left value is bigger     │ 5 > 3   │ True   │
  #│ <        │ Less than                │ Checks if left value is smaller    │ 5 < 3   │ False  │
  #│ >=       │ Greater than or equal to │ Checks if left is bigger or equal  │ 5 >= 5  │ True   │
  #│ <=       │ Less than or equal to    │ Checks if left is smaller or equal │ 5 <= 3  │ False  │
  #└──────────┴──────────────────────────┴────────────────────────────────────┴─────────┴────────┘
```

```python
  #4. Assignment Operators
  #Used to assign/store values into variables. They also come in shorthand (compound) forms to write shorter code.

  #┌──────────┬───────────┬─────────────────────┬───────────────────────┐
  #│ Operator │ Same as   │ Simple Definition   │ Example               │
  #├──────────┼───────────┼─────────────────────┼───────────────────────┤
  #│ =        │ x = 5     │ Direct Assignment   │ x = 5 (stores 5 in x) │
  #│ +=       │ x = x + 5 │ Add and Assign      │ x += 5                │
  #│ -=       │ x = x - 5 │ Subtract and Assign │ x -= 5                │
  #│ *=       │ x = x * 5 │ Multiply and Assign │ x *= 5                │
  #│ /=       │ x = x / 5 │ Divide and Assign   │ x /= 5                │
  #│ %=       │ x = x % 5 │ Modulus and Assign  │ x %= 5                │
  #└──────────┴───────────┴─────────────────────┴───────────────────────┘
```

```python
#Simple Intrest or Compounnd Intrest

# PG 174 , Q.11 (practicle of half yearly)
# Simple Interest
p = float(input("Enter Principle Amount: "))
r = float(input("Enter Rate of Interest: "))
t = float(input("Enter Time in years: "))

# Fixed: added '*' between variables
si = (p * r * t) / 100
print("The Simple Interest is:", si)

print("-" * 30)

# Compound Interest
p = float(input("Enter the Principle Amount: "))
r = float(input("Enter the Rate Of Interest (in %): "))
t = float(input("Enter Time in years: "))  # Fixed: added missing time variable
n = float(input("Enter the number of times compounded per year: "))

# Formula expects rate as a decimal (e.g., 5% -> 0.05)
r_decimal = r / 100

ci = p * (1 + (r_decimal / n)) ** (n * t)

print("The Compound Interest amount is:", ci)
```
